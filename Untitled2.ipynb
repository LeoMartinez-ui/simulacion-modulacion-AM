```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

print("Librerías cargadas correctamente")
```

```python
# Parámetros de la señal
fm = 5                 # Frecuencia del mensaje (Hz)
fs = 1000              # Frecuencia de muestreo
t = np.arange(0, 1, 1/fs)

# Crear señal de información
mensaje = np.sin(2 * np.pi * fm * t)

# Graficar señal original
plt.figure(figsize=(10, 4))
plt.plot(t, mensaje)
plt.title("Señal original de información")
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Amplitud")
plt.grid()
plt.show()
```

```python
# Parámetros de la portadora
fc = 100              # Frecuencia de la portadora (Hz)
Ac = 1                # Amplitud de la portadora
m = 0.8               # Índice de modulación

# Crear señal portadora
portadora = Ac * np.cos(2 * np.pi * fc * t)

# Aplicar modulación AM
senal_am = Ac * (1 + m * mensaje) * np.cos(2 * np.pi * fc * t)

# Graficar señal modulada
plt.figure(figsize=(12, 5))
plt.plot(t, senal_am)
plt.title("Señal modulada en amplitud (AM)")
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Amplitud")
plt.grid()
plt.show()
```

```python
# Calcular la transformada de Fourier
N = len(senal_am)
frecuencias = fftfreq(N, 1/fs)
espectro = np.abs(fft(senal_am)) / N

# Graficar espectro de frecuencias
plt.figure(figsize=(12, 5))
plt.plot(frecuencias[:N//2], espectro[:N//2])
plt.title("Espectro de frecuencias de la señal AM")
plt.xlabel("Frecuencia (Hz)")
plt.ylabel("Amplitud")
plt.grid()
plt.show()
```

```python
# Generar ruido aleatorio
ruido = np.random.normal(0, 0.3, len(senal_am))

# Agregar ruido a la señal AM
senal_ruidosa = senal_am + ruido

# Graficar comparación
plt.figure(figsize=(12, 5))
plt.plot(t, senal_am, label="Señal AM original")
plt.plot(t, senal_ruidosa, label="Señal AM con ruido", alpha=0.7)
plt.title("Comparación de señal AM con ruido")
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Amplitud")
plt.legend()
plt.grid()
plt.show()
```

```python
# Aplicar atenuación del 50%
factor_atenuacion = 0.5
senal_atenuada = senal_am * factor_atenuacion

# Graficar comparación
plt.figure(figsize=(12, 5))
plt.plot(t, senal_am, label="Señal AM original")
plt.plot(t, senal_atenuada, label="Señal atenuada")
plt.title("Efecto de la atenuación en la señal AM")
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Amplitud")
plt.legend()
plt.grid()
plt.show()
```

```python
# Simular distorsión mediante una componente adicional
senal_distorsionada = senal_am + 0.3 * np.sin(2 * np.pi * 200 * t)

# Graficar comparación
plt.figure(figsize=(12, 5))
plt.plot(t, senal_am, label="Señal AM original")
plt.plot(t, senal_distorsionada, label="Señal distorsionada")
plt.title("Efecto de la distorsión en la señal AM")
plt.xlabel("Tiempo (segundos)")
plt.ylabel("Amplitud")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(14, 10))

plt.subplot(3, 1, 1)
plt.plot(t, senal_am)
plt.title("1. Señal AM original")
plt.grid()

plt.subplot(3, 1, 2)
plt.plot(t, senal_ruidosa)
plt.title("2. Señal AM con ruido")
plt.grid()

plt.subplot(3, 1, 3)
plt.plot(t, senal_atenuada)
plt.title("3. Señal AM con atenuación")
plt.xlabel("Tiempo (segundos)")
plt.grid()

plt.tight_layout()
plt.show()
```
